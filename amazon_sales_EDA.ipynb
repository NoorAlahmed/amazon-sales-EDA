## processsing and cleaning the dataset
data cleaning is an important step to ensure the dataset is accurate, consistent, and ready for analysis.

```python
#changing the data type of discounted price and actual price
"""these columns were originally stored as text (because of currency signs), which prevents proper analysis
so symbols are removed to allow numerical operations and visualizations."""
df['discounted_price'] = df['discounted_price'].str.replace("₹",'')
df['discounted_price'] = df['discounted_price'].str.replace(",",'')
df['discounted_price'] = df['discounted_price'].astype('float64')

df['actual_price'] = df['actual_price'].str.replace("₹",'')
df['actual_price'] = df['actual_price'].str.replace(",",'')
df['actual_price'] = df['actual_price'].astype('float64')
```

```python
# Changing Datatype and values in Discount Percentage

df['discount_percentage'] = df['discount_percentage'].str.replace('%','').astype('float64')

df['discount_percentage'] = df['discount_percentage'] / 100
```

```python
# Finding unusual string in rating column
df['rating'].value_counts()
```

```python
# Check the strange row
df.query('rating == "|"')
```

```python
# Changing Rating Columns Data Type

df['rating'] = df['rating'].str.replace('|', '3.9').astype('float64')
# replaced it with 3.9 as it was in the orginal file
```

```python
# Changing 'rating_count' Column Data Type

df['rating_count'] = df['rating_count'].str.replace(',', '').astype('float64')
```

**TASK 3: Detect and handle outliers using IQR**

Detect Outliers in Price Columns

```python
numerical_cols = ['actual_price', 'discounted_price',
                  'rating', 'rating_count',
                  'discount_percentage']

for col in numerical_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col}: {len(outliers)} outliers")
```

Visualize Outliers

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df[numerical_cols])
plt.title("Boxplots Before Outlier Treatment")
plt.show()
```

Handle Outliers

```python
for col in numerical_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = np.where(df[col] > upper,
                       upper,
                       np.where(df[col] < lower,
                                lower,
                                df[col]))
```

Show the result

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df[numerical_cols])
plt.title("Boxplots After Outlier Treatment")
plt.show()
```

**Task 4: Normalize Numerical Features using Min-Max and Z-score**

Min-Max Normalization

```python
minmax_scaler = MinMaxScaler()

df_minmax = df.copy()

df_minmax[numerical_cols] = minmax_scaler.fit_transform(
    df[numerical_cols]
)

df_minmax[numerical_cols].head()
print("Data after Min-Max Scaling:")
print(df_minmax[numerical_cols].head())
print("Min-Max scaling preserves the shape of the original distribution while transforming all values into the range 0 to 1.")
```

Z-Score Standardization

```python
zscore_scaler = StandardScaler()

df_zscore = df.copy()

df_zscore[numerical_cols] = zscore_scaler.fit_transform(
    df[numerical_cols]
)

df_zscore[numerical_cols].head()
```

**Compare Distributions**

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(df_minmax['actual_price'], bins=30)
plt.title("Min-Max Scaled")

plt.subplot(1,2,2)
plt.hist(df_zscore['actual_price'], bins=30)
plt.title("Z-Score Scaled")

plt.show()
print("Min-Max scaling transforms the data to a fixed range (0 to 1), while Z-Score scaling standardizes the data to have a mean of 0 and a standard deviation of 1, which can be more effective for algorithms that assume normally distributed data.")
```

**Task 5: Apply PCA and Interpret Explained Variance**

```python
features = ['actual_price',
            'discounted_price',
            'discount_percentage',
            'rating',
            'rating_count']

X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#Apply PCA
pca = PCA()
X_pca = pca.fit_transform(X_scaled)
```

```python
explained_variance = pca.explained_variance_ratio_

print(explained_variance)
```

**PCA Explained Variance Interpretation**

The PCA results show that:

PC1 explains 41.15% of the variance.
PC2 explains 23.08% of the variance.
PC3 explains 19.26% of the variance.
PC4 explains 15.84% of the variance.
PC5 explains 0.67% of the variance.

Together, the first three principal components explain approximately 83.99% of the total variance,indicating that they capture most of the information in the original dataset.
This suggests that PCA can effectively reduce the dimensionality of the dataset while preserving the majority of its information.

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(1, len(explained_variance)+1),
    np.cumsum(explained_variance),
    marker='o'
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA Explained Variance")

plt.grid(True)
plt.show()
```

Interpretation: With 3 components preserving 84% of variance, we can reduce from 5 features to 3 while keeping most information.
